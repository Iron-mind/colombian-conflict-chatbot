```python
from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd

```

```python

folder_path = '/content/drive/My Drive/Univalle/tesis/Tesis/reports-pdf'

file_list = []
for f in os.listdir(folder_path):
    name, ext = os.path.splitext(f)
    ext = ext[1:] if ext else ''
    file_list.append({'name': name, 'extension': ext})

df = pd.DataFrame(file_list)

with open('/content/en_drive.json', 'w') as f:
    df.to_json(f, orient='records')
```

```python
import json
with open('/content/en_local.json') as f:
    local = json.load(f)
with open('/content/en_drive.json') as f:
    drive = json.load(f)

local_names = {item['name'] + '.' + item['extension'] for item in local}
drive_names = {item['name'] + '.' + item['extension'] for item in drive}
missing = sorted(local_names - drive_names)

for i in missing:
  print(i)
```
