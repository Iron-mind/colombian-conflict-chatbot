```python
import pandas as pd
import glob
import os

# Definir los nuevos nombres de las columnas
nuevos_nombres = [
    "Exactitud de la información",
    "Relevancia con respecto a la pregunta",
    "Cobertura del contenido",
    "Claridad de redacción y comprensión",
    "Justificación o respaldo con evidencia",
    "Evaluación general",
    "Comentarios adicionales (opcional)"
]

# Carpeta donde se encuentran los archivos Excel (cambia esto por tu ruta)
LISTING_PATH = "/content/drive/My Drive/Univalle/tesis/Tesis/resources/respuestas generadas y evaluadas"

carpeta = LISTING_PATH  # Reemplaza con la ruta correcta

df_combinado = pd.DataFrame()
archivos_excel = glob.glob(os.path.join(carpeta, '*.xlsx')) + glob.glob(os.path.join(carpeta, '*.xls'))

for archivo in archivos_excel:
    print(f"Procesando {archivo}")
    try:
        df = pd.read_excel(archivo, header=None)

        # Verificar que el archivo tenga suficientes columnas
        if df.shape[1] < 11:
            print(f"Advertencia: {archivo} no tiene suficientes columnas. Se omitirá.")
            continue

        datos = df.iloc[1:, 4:11]  # Filas desde la 2da, columnas E-K
        datos.columns = nuevos_nombres
        df_combinado = pd.concat([df_combinado, datos], ignore_index=True)

    except Exception as e:
        print(f"Error al procesar {archivo}: {str(e)}")
        continue

# Verificar que el DataFrame no esté vacío
if not df_combinado.empty:
    columnas_numericas = nuevos_nombres[:-1]

    # Verificar que las columnas existan antes de convertirlas
    columnas_existentes = [col for col in columnas_numericas if col in df_combinado.columns]

    if columnas_existentes:
        df_combinado[columnas_existentes] = df_combinado[columnas_existentes].apply(pd.to_numeric, errors='coerce')

        # Calcular estadísticas
        estadisticas = pd.DataFrame({
            'Promedio': df_combinado[columnas_existentes].mean(),
            'Desviación estándar': df_combinado[columnas_existentes].std()
        })

        print("DataFrame combinado:")
        print(df_combinado.head())
        print("\nEstadísticas:")
        print(estadisticas)
    else:
        print("No se encontraron las columnas numéricas esperadas.")
else:
    print("No se pudieron cargar datos de ningún archivo.")
```

```python
from datasets import load_dataset
from transformers import AutoTokenizer

# Carga tu dataset de Hugging Face
dataset = load_dataset("jdavit/colombian-conflict-SQA")  # Reemplaza con el nombre de tu dataset

# Elige un tokenizador adecuado (por ejemplo, 'bert-base-uncased' o el que prefieras)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
def calculate_stats(dataset, tokenizer, split="train"):
    # Inicializa acumuladores
    total_questions = 0
    total_answers = 0
    total_contexts = 0
    sum_question_tokens = 0
    sum_answer_tokens = 0
    sum_context_tokens = 0

    # Itera sobre el dataset
    for example in dataset[split]:
        # Tokeniza y cuenta los tokens
        question_tokens = len(tokenizer.tokenize(example["question"]))
        answer_tokens = len(tokenizer.tokenize(example["answer"]))
        context_tokens = len(tokenizer.tokenize(example["context"]))

        # Acumula
        sum_question_tokens += question_tokens
        sum_answer_tokens += answer_tokens
        sum_context_tokens += context_tokens
        total_questions += 1
        total_answers += 1
        total_contexts += 1

    # Calcula promedios
    avg_question = sum_question_tokens / total_questions
    avg_answer = sum_answer_tokens / total_answers
    avg_context = sum_context_tokens / total_contexts

    return {
        "avg_question_tokens": avg_question,
        "avg_answer_tokens": avg_answer,
        "avg_context_tokens": avg_context
    }

# Ejemplo de uso (asumiendo que el dataset tiene un split "train")
stats = calculate_stats(dataset, tokenizer, split="train")
print(stats)
```

```python

# 2. Función para calcular rangos (min, max)
def calculate_ranges(dataset, tokenizer, split="train"):
    # Inicializa variables para rangos
    question_lengths = []
    answer_lengths = []
    context_lengths = []

    # Procesa ejemplos (usa .select() para muestras grandes)
    for example in dataset[split]:
        question_lengths.append(len(tokenizer.tokenize(example["question"])))
        answer_lengths.append(len(tokenizer.tokenize(example["answer"])))
        context_lengths.append(len(tokenizer.tokenize(example["context"])))

    # Calcula estadísticas
    stats = {
        "preguntas": {
            "min_tokens": min(question_lengths),
            "max_tokens": max(question_lengths),
            "avg_tokens": sum(question_lengths) / len(question_lengths)  # Opcional: promedio
        },
        "respuestas": {
            "min_tokens": min(answer_lengths),
            "max_tokens": max(answer_lengths),
            "avg_tokens": sum(answer_lengths) / len(answer_lengths)  # Opcional
        },
        "contextos": {
            "min_tokens": min(context_lengths),
            "max_tokens": max(context_lengths),
            "avg_tokens": sum(context_lengths) / len(context_lengths)  # Opcional
        }
    }
    return stats

# 3. Ejecuta y muestra resultados
stats = calculate_ranges(dataset, tokenizer, split="train")
print("📊 Estadísticas de Tokens:")
for key in stats:
    print(f"\n🔹 {key.capitalize()}:")
    print(f"   - Mínimo: {stats[key]['min_tokens']} tokens")
    print(f"   - Máximo: {stats[key]['max_tokens']} tokens")
    print(f"   - Promedio: {stats[key]['avg_tokens']:.1f} tokens")  # Opcional
```
