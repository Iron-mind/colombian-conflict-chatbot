```python
from huggingface_hub import login
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer

# Autenticación (usa tu token)
login(token="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")

# Cargar modelo base y PEFT
base_model = AutoModelForCausalLM.from_pretrained("mistralai/Mistral-7B-Instruct-v0.3")
model = PeftModel.from_pretrained(base_model, "raulgdp/Mistral-7B-Instruct-v0.3-JEP")

# Cargar tokenizador
tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-Instruct-v0.3")

# Generar respuesta
prompt = "<s>[INST] Explica qué es el aprendizaje por transferencia. [/INST]"
inputs = tokenizer(prompt, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=200)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
